```python
path2='/projekt1/ag_maahn/data_obs_nobackup/amazonas/cover2000/'

kacheln2 = [
    rxr.open_rasterio(path2+"Hansen_GFC-2024-v1.12_treecover2000_00N_050W.tif"),
    rxr.open_rasterio(path2+"Hansen_GFC-2024-v1.12_treecover2000_00N_060W.tif"),
    rxr.open_rasterio(path2+"Hansen_GFC-2024-v1.12_treecover2000_00N_070W.tif"),
    rxr.open_rasterio(path2+"Hansen_GFC-2024-v1.12_treecover2000_00N_080W.tif"),
    rxr.open_rasterio(path2+"Hansen_GFC-2024-v1.12_treecover2000_10N_050W.tif"),
    rxr.open_rasterio(path2+"Hansen_GFC-2024-v1.12_treecover2000_10N_060W.tif"),
    rxr.open_rasterio(path2+"Hansen_GFC-2024-v1.12_treecover2000_10N_070W.tif"),
    rxr.open_rasterio(path2+"Hansen_GFC-2024-v1.12_treecover2000_10N_080W.tif"),
    rxr.open_rasterio(path2+"Hansen_GFC-2024-v1.12_treecover2000_10S_050W.tif"),
    rxr.open_rasterio(path2+"Hansen_GFC-2024-v1.12_treecover2000_10S_060W.tif"),
    rxr.open_rasterio(path2+"Hansen_GFC-2024-v1.12_treecover2000_10S_070W.tif"),
    rxr.open_rasterio(path2+"Hansen_GFC-2024-v1.12_treecover2000_10S_080W.tif")
]

mrd2 =merge_arrays(kacheln2)
```

```python
loss = mrd1.to_dataset(name="lossyear")
cvr = mrd2.to_dataset(name="cover2000")
ds=xr.merge([loss, cvr])
ds
```

```python
# set boundary values for map-plots
lon_min, lon_max = -85, -35
lat_min, lat_max = -35, 15
```

```python
th=50 # % Threshold for beeing marked as forest, following the definition in the dataset for forest loss

maxX = round(float(ds['x'].max()),0)
minX = round(float(ds['x'].min()),0)

maxY = round(float(ds['y'].max()),0)
minY = round(float(ds['y'].min()),0)
 
#x_range = np.arange(minX, maxX+1)
#y_range = np.arange(minY, maxY+1)
#print(x_range)
#print(y_range)
```

```python
# 1. create bolean arrays
cond_forest = (ds['cover2000'] >= th) & ((ds['lossyear'] >= 2) | (ds['lossyear'] == 0))
cond_no_loss = ds['lossyear'] <= 2
# 2. Group in 1x1° bins and sum
forest_count      = cond_forest.groupby_bins('x', np.arange(minX, maxX+1)) \
                               .sum() \
                               .groupby_bins('y', np.arange(minY, maxY+1)) \
                               .sum()

still_forest_count = (cond_forest & cond_no_loss) \
                               .groupby_bins('x', np.arange(minX, maxX+1)) \
                               .sum() \
                               .groupby_bins('y', np.arange(minY, maxY+1)) \
                               .sum()

# 3. Degree of Deforestation
DoD = (forest_count - still_forest_count) / forest_count.where(forest_count > 0)

#Difference (pixel count) of forest area in 2001 and 2024

Fo_diff=forest_count.where(forest_count > 0) - still_forest_count
```

```python
ds_dod = DoD.to_dataset(name="DoD")
ds_Fo_diff= Fo_diff.to_dataset(name="Difference").squeeze()

maxdef=int(Fo_diff.max())
```

```python
x_mids = np.array([iv.mid for iv in ds_Fo_diff.x_bins.values])
y_mids = np.array([iv.mid for iv in ds_Fo_diff.y_bins.values])

fig, ax = plt.subplots(figsize=(12, 8), subplot_kw={'projection': ccrs.PlateCarree()})

ds_Fo_diff['Difference'].plot.pcolormesh(
    ax=ax,
    transform=ccrs.PlateCarree(),
    cmap='RdYlGn_r',
    vmin=0, vmax=maxdef,
    cbar_kwargs={'label': 'Difference of Forrest states between 2001 and 2024 \n (pixel count)'}
)

ax.set_extent([lon_min, lon_max, lat_min, lat_max], crs=ccrs.PlateCarree())

ax.add_feature(cfeature.COASTLINE, linewidth=0.8)
ax.add_feature(cfeature.BORDERS, linewidth=0.5)
ax.add_feature(cfeature.RIVERS, linewidth=0.5)

gl = ax.gridlines(draw_labels=True)
gl.xlocator = mticker.FixedLocator(range(-90, -39, 10))
gl.ylocator = mticker.FixedLocator(range(-30, 11, 10))

ax.set_title('Forest difference 2000–2024')
plt.tight_layout()
plt.savefig('/home/oscholz/plots/forest/difference.png')
plt.show()
```

# Modis Datensatz

```python
def plot_map(Data, col_min, col_max, sig_data, colormap="RdBu_r", typ=None, sig=None, sy=None, ey=None, season=None, data=None, header=None, folder=None):
    fig, ax = plt.subplots(figsize=(12, 8), subplot_kw={"projection": ccrs.PlateCarree()})

    was=" ".join(header.split()[:3])
    if sig:
        label='trend (per decade)'
        name='trend'
    else:
        label='p-value'
        name='pval'
 
    if typ=='mean_avg':
        label=f'mean {was}'
        name='clim_Mean'
    
    (Data).plot.pcolormesh(
        ax=ax,
        transform=ccrs.PlateCarree(),
        cmap=colormap,
        vmin=col_min,   
        vmax=col_max, 
        cbar_kwargs={"label": label}
    )

    if sig :
    # Non-significant areas hatched (p > 0.05)
        sig_mask = sig_data > 0.05
        ax.contourf(
            Data.longitude, Data.latitude,
            sig_mask.values,
            levels=[0.5, 1.5],
            hatches=["////"],
            colors="none",
            transform=ccrs.PlateCarree()
        )
    
    ax.add_feature(cfeature.COASTLINE, linewidth=0.8)
    ax.add_feature(cfeature.BORDERS, linewidth=0.5)
    ax.set_extent([lon_min, lon_max, lat_min, lat_max], crs=ccrs.PlateCarree())
    
    gl = ax.gridlines(draw_labels=True)
    gl.xlocator = mticker.FixedLocator(range(-90, -39, 10))
    gl.ylocator = mticker.FixedLocator(range(-30, 11, 10))

    if sy=='2002' and ey=='2026':
        f2='/'
    else:
        f2='/var_years/'   
    ax.set_title(f"Trend {header} {sy}–{ey}\n(Hatching = not significant, p > 0.05)")
    savepath=f'/home/oscholz/plots/{folder}{f2}{name}_{data}_{sy}_{ey}.png'
    plt.savefig(savepath)
    plt.tight_layout()
    plt.show()

#---------------------------------------------------------------------------------------------------------------------------------------------------
#---------------------------------------------------------------------------------------------------------------------------------------------------

def trends (Dataset):
    time_num = (Dataset.time - Dataset.time[0]).dt.days.values
    
    def linear_trend_with_pvalue(y):
        mask = ~np.isnan(y)
        if mask.sum() < 2:
            return np.nan, np.nan
        slope, _, _, pvalue, _ = stats.linregress(time_num[mask], y[mask])
        return slope, pvalue
        
    # Slope und p-Wert berechnen
    slope, pvalue = xr.apply_ufunc(
        linear_trend_with_pvalue,
        Dataset["Mean"],
        input_core_dims=[["time"]],
        output_core_dims=[[], []],
        vectorize=True,
        output_dtypes=[float, float]
    )
    
    trend_modis = xr.Dataset({
        "trend_Mean": slope.transpose("latitude", "longitude"),
        "pvalue":     pvalue.transpose("latitude", "longitude")
    })
    trend_modis['trend_Mean']=trend_modis['trend_Mean']*365*10
    return trend_modis
    
#---------------------------------------------------------------------------------------------------------------------------------------------------
#---------------------------------------------------------------------------------------------------------------------------------------------------

def bivariate_maps(data_modis, only_sig, data=None, header=None, folder=None):
    
# ── 1. Interpolate Modis Data to Hansen-Grid ──────────────────────────────────
    if only_sig:
        ols=0.05
    else:
        ols=1
    trend_sig = data_modis["trend_Mean"].where(data_modis["pvalue"] < ols)
    
    x_mids = np.array([iv.mid for iv in ds_dod.x_bins.values])
    y_mids = np.array([iv.mid for iv in ds_dod.y_bins.values])
    
    trend_interp = trend_sig.interp(
        latitude=y_mids,
        longitude=x_mids,
        method="nearest"
    )
    
    # ── 2. extract values──────────────────────────────────────────────────────
    fodif_vals = ds_Fo_diff["Difference"].squeeze().values # (30, 40)
    trend_vals = trend_interp.values                       # (30, 40)
    
    # ── 3. classification ─────────────────────────────────────────────────────────
    both_valid_fo = ~np.isnan(fodif_vals) & ~np.isnan(trend_vals)
    fodif_t = np.nanpercentile(fodif_vals[both_valid_fo], [33, 66])
    
    # Trend: 4 classes with 0 as natürlicher threshold
    neg_vals = trend_vals[both_valid_fo & (trend_vals < 0)]
    pos_vals = trend_vals[both_valid_fo & (trend_vals > 0)]
    trend_t = [np.median(neg_vals), 0, np.median(pos_vals)]
    # → 0: stark neg | 1: schwach neg | 2: schwach pos | 3: stark pos
    
    def classify_deforestation(arr, thresholds):
        out = np.full(arr.shape, np.nan)
        out[arr <= thresholds[0]] = 0
        out[(arr > thresholds[0]) & (arr <= thresholds[1])] = 1
        out[arr > thresholds[1]] = 2
        return out
        
    def classify_trend(arr, thresholds):
        out = np.full(arr.shape, np.nan)
        out[arr <= thresholds[0]] = 0
        out[(arr > thresholds[0]) & (arr <= thresholds[1])] = 1
        out[(arr > thresholds[1]) & (arr <= thresholds[2])] = 2
        out[arr > thresholds[2]] = 3
        return out
    
    trend_class = classify_trend(trend_vals, trend_t)
    dof_class = classify_deforestation(fodif_vals, fodif_t)
    
    # combined Index 0–11 (3 DoD × 4 Trend)
    combo_fo = np.full(fodif_vals.shape, np.nan)
    combo_fo[both_valid_fo] = dof_class[both_valid_fo] * 4 + trend_class[both_valid_fo]
    
    # ── 4. Bivariate colour matrix (3×4) ───────────────────────────────────────────
    # Zeilen = DoD (0=niedrig → 2=hoch), Spalten = Trend (0=stark neg → 3=stark pos)
    bivar_colors = [
        "lightblue",   # DoD low,  Trend stark neg
        "lightcyan",   # DoD low,  Trend schwach neg
        "lightyellow", # DoD low,  Trend schwach pos
        "yellow",      # DoD low,  Trend stark pos
        "blue",        # DoD mid,  Trend stark neg
        "cornflowerblue", # DoD mid, Trend schwach neg
        "orange",      # DoD mid,  Trend schwach pos
        "darkorange",  # DoD mid,  Trend stark pos
        "darkblue",    # DoD high, Trend stark neg
        "purple",      # DoD high, Trend schwach neg
        "tomato",      # DoD high, Trend schwach pos
        "red",         # DoD high, Trend stark pos
    ]
    cmap_bivar = ListedColormap(bivar_colors)
    
    #plot map with forrest difference ---------------------------------------------------------------------------------------------------------- 
    
    fig, ax = plt.subplots(
        figsize=(12, 9),
        subplot_kw={"projection": ccrs.PlateCarree()}
    )
    ax.set_extent([lon_min, lon_max, lat_min, lat_max], crs=ccrs.PlateCarree())
    
    ax.add_feature(cfeature.COASTLINE, linewidth=0.8)
    ax.add_feature(cfeature.BORDERS,   linewidth=0.5)
    
    #Forest Difference as background data
    
    im = ax.pcolormesh(
        x_mids, y_mids, combo_fo,
        cmap=cmap_bivar, vmin=0, vmax=11,
        transform=ccrs.PlateCarree(), zorder=2
    )
    
    # Mask for NaN (not significant or no forrest)
    nan_mask = np.where(np.isnan(combo_fo), ds_Fo_diff['Difference'], np.nan)
    im = ax.pcolormesh(
        x_mids, y_mids, nan_mask,
        cmap='Greys', vmin=0, vmax=int(Fo_diff.max()),
        transform=ccrs.PlateCarree(), zorder=1
    )
    
    cbar = fig.colorbar(im, ax=ax, orientation='vertical', pad=0.05, shrink=0.7)
    cbar.set_label('Forrest Area Difference between 2001 and 2024 \n where trend is not significant (pixel count)')
    ax.gridlines(draw_labels=True, linewidth=0.3, color="gray", alpha=0.5)
    ax.set_title("Bivariate map: \n"
                 f"'Forrest Area Difference 2001 and 2024 & Trend of {header}\n"
                 "(only significant trends, p < 0.05)", fontsize=13)
    
    # ── 6. Bivariate legend  ──────────────────────────────────────
    white_box = FancyBboxPatch(
        (0.015, 0.015), 0.35, 0.2,          
        transform=ax.transAxes,
        boxstyle="square,pad=0",
        facecolor="white", edgecolor="none",
        zorder=5                           
    )
    ax.add_patch(white_box)
    legend_ax = fig.add_axes([0.22 , 0.15, 0.16, 0.12])
    legend_grid = np.arange(12).reshape(3, 4)
    legend_ax.imshow(legend_grid, cmap=cmap_bivar, vmin=0, vmax=11,
                     origin="lower", aspect="auto")
    legend_ax.set_xticks([0, 1, 2, 3])
    legend_ax.set_xticklabels(["↓↓ Trend", "↓ Trend", "↑ Trend", "↑↑ Trend"], fontsize=8)
    legend_ax.set_yticks([0, 1, 2])
    legend_ax.set_yticklabels(["DoD\nniedrig", "mittel", "hoch"], fontsize=10)
    legend_ax.set_title("Legende", fontsize=10)
    legend_ax.tick_params(length=0)
    
    #plt.tight_layout()
    savepath2=f'/home/oscholz/plots/{folder}/bivariate_map_area_{data}.png'
    #plt.savefig(savepath2)
    plt.show()

#---------------------------------------------------------------------------------------------------------------------------------------------------
#---------------------------------------------------------------------------------------------------------------------------------------------------

def analyze_modis(dataset, typ, season , start='2002-07-04', end  ='2026-03-11',col_min=None, col_max=None, colormap="RdBu_r",colormap_mean='turbo', only_sig=True, biv=True):
    
    #season choose between: 
    #'off' for a total annual view       or
    #dry for dry season analysis         or
    #wet for wet season analysis
    
    #dataset choose between: 
    # 'COT_liq.nc' or    (Cloud Optical Thickness of liqiud clouds)
    # 'CMF_low.nc' or    (Cloud Mask Fraction of low clouds) 
    # 'CPS_liq.nc'       (Cloud Particle Size of liquid cloud)
    #or create a new file with concat.ipynb and choose that one

    #typ choose between:
    #'mean_avg'
    #'trend_Mean'
    #'pvalue'
    
    signif=False if typ=='pvalue' or typ=='mean_avg' else True
    modis=xr.open_dataset('/projekt1/ag_maahn/data_obs_nobackup/modis/MCD06COSP_D3/merged/'+dataset)
    
    sy = start[:4]
    ey = end[:4]
    
    if dataset=='COT_liq.nc':
        folder='cot_liq'
        data=folder
        header='Cloud Optical Thickness of liquid clouds' 
        cmin=None
        cmax=None
    elif dataset=='CMF_low.nc':
        folder='cmf_low'
        data=folder
        header='Cloud Mask Fraction of low clouds'
        cmin=-0.05
        cmax=0.05
    elif dataset=='CPS_liq.nc':
        folder='cps_liq'
        data=folder
        header='Cloud Particle Size of liquid clouds'
        cmin=None
        cmax=None
    
    if season=='dry':
        modis = modis.sel(time=modis.time.dt.month.isin([6,7,8,9]))
        header=header+' in dry season'
        data=folder+'_dry'
    elif season=='wet':
        modis = modis.sel(time=~modis.time.dt.month.isin([6,7,8,9]))
        header=header+' in wet season'
        data=folder+'_wet'
    else:
        pass

    modis=modis.sel(time=slice(start, end))
    modis['mean_avg']=modis['Mean'].mean(dim='time', skipna=True).transpose("latitude", "longitude")

    trend_modis=trends(modis)

    if typ=='mean_avg':
        data2=modis['mean_avg']
        cmin=0
        cmax=1
        colormap='turbo'
    elif typ=='pvalue':
        cmin=0
        cmax=0.1
        data2=trend_modis[typ]
    else:
        data2=trend_modis[typ]

    plot_map(data2, col_min=cmin, col_max=cmax, sig_data=trend_modis["pvalue"], colormap=colormap,typ=typ, sig=signif, sy=sy, ey=ey, season=season, data=data, header=header, folder=folder)
    if biv:
        bivariate_maps(trend_modis,data=data, header=header, folder=folder, only_sig=only_sig)
```

```python
# ── Pixelfläche in km² pro 1°×1°-Box ─────────────────────────────────────────
R = 6371.0  # Erdradius in km
delta_lambda = np.deg2rad(1.0)  # 1° in Längenrichtung

# y_mids sind Mittelpunkte der Boxen → Grenzen bei ±0.5°
phi1 = np.deg2rad(y_mids - 0.5)
phi2 = np.deg2rad(y_mids + 0.5)
box_area_km2 = R**2 * delta_lambda * np.abs(np.sin(phi2) - np.sin(phi1))  # (n_lat,)

# Auf alle Längengrade ausdehnen: (n_lat,) → (n_lat, n_lon)
box_area_km2_2d = np.tile(box_area_km2[:, np.newaxis], (1, len(x_mids)))

# Pixelfläche in km²
pixels_per_box = 4000 * 4000
pixel_area_km2 = box_area_km2_2d / pixels_per_box  # (n_lat, n_lon)

# Amazonas-Maske anwenden
pixel_area_amz = pixel_area_km2[lat_mask, :]

# Fo_diff in km² umrechnen
dof_km2_vals_amz = dof_vals_amz * pixel_area_amz  # elementweise

# ── Daten für Regression ──────────────────────────────────────────────────────
dof_km2_flat  = dof_km2_vals_amz[both_valid_amz2].flatten()
# trend_flat2 bleibt gleich (gleiche Maske)

slope3, intercept3, r_value3, p_value3, std_err3 = stats.linregress(
    dof_km2_flat / 1000, trend_flat2  # pro 1000 km² für lesbare Slope-Werte
)

# ── Plot 3: Fo_diff in km² × Trend ───────────────────────────────────────────
fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(dof_km2_flat / 1000, trend_flat2, alpha=0.6, edgecolors="k",
           linewidths=0.3, s=40, color="steelblue")
x_line3 = np.linspace((dof_km2_flat / 1000).min(), (dof_km2_flat / 1000).max(), 100)
y_line3 = slope3 * x_line3 + intercept3
ax.plot(x_line3, y_line3, color="red", linewidth=2,
        label=f"Slope: {slope3:.4f}\n$R^2$ = {r_value3**2:.3f}\np = {p_value3:.4f}")
ax.axhline(0, color="gray", linewidth=0.8, linestyle="--")
ax.set_xlabel("Forest loss (1000 km²)", fontsize=12)
ax.set_ylabel(f"{header} (per decade)", fontsize=12)
ax.set_title(f"Forest loss (km²) & {header} Trend (low clouds)\n"
             "Amazona(14°S–7°N, 80°W-40°W), only significant trends (p < 0.05)", fontsize=13)
ax.legend(fontsize=11)
plt.tight_layout()
plt.savefig(f'/home/oscholz/plots/{folder}/scatter_area_{data}.png', dpi=150, bbox_inches="tight")
plt.show()
```

## cross correlation because of trade winds 

```python
# ── Räumliche Kreuzkorrelation: Dof vs. Trend ─────────────────────────────────
# Versatz-Bereich: ±5 Gitterzellen (= ±5°)
shift_range = np.arange(-5, 6)  # in Gitterzellen (1° pro Zelle)
r_spatial = np.full((len(shift_range), len(shift_range)), np.nan)

for i, d_lat in enumerate(shift_range):
    for j, d_lon in enumerate(shift_range):
        # DoF-Feld verschieben
        dof_shifted = np.roll(np.roll(dof_vals_amz, d_lat, axis=0), d_lon, axis=1)

        # Randeffekte durch np.roll maskieren
        mask = np.ones_like(dof_vals_amz, dtype=bool)
        if d_lat > 0:
            mask[:d_lat, :] = False
        elif d_lat < 0:
            mask[d_lat:, :] = False
        if d_lon > 0:
            mask[:, :d_lon] = False
        elif d_lon < 0:
            mask[:, d_lon:] = False

        valid = mask & ~np.isnan(dof_shifted) & ~np.isnan(trend_vals_amz)
        if valid.sum() < 10:  # zu wenige Punkte
            continue

        r, p = stats.pearsonr(dof_shifted[valid], trend_vals_amz[valid])
        r_spatial[i, j] = r if p < 0.05 else np.nan

# ── Plot: Korrelations-Heatmap ────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(8, 7))
im = ax.imshow(r_spatial, cmap="RdBu_r", vmin=-1, vmax=1,
               origin="lower", aspect="auto")
plt.colorbar(im, ax=ax, label="Pearson r (nur p < 0.05)")

ax.set_xticks(np.arange(len(shift_range)))
ax.set_xticklabels([f"{s:+d}°" for s in shift_range])
ax.set_yticks(np.arange(len(shift_range)))
ax.set_yticklabels([f"{s:+d}°" for s in shift_range])
ax.set_xlabel("shifted longitude ", fontsize=12)
ax.set_ylabel("shifted latitude ", fontsize=12)
ax.axhline(len(shift_range)//2, color="k", linewidth=0.8, linestyle="--")
ax.axvline(len(shift_range)//2, color="k", linewidth=0.8, linestyle="--")
ax.set_title("cross correlation: Cloud Trend & Difference in forrest area", fontsize=12)

# Besten Versatz markieren
best_idx = np.unravel_index(np.nanargmax(np.abs(r_spatial)), r_spatial.shape)
best_idx2 = np.unravel_index(np.nanargmin(r_spatial), r_spatial.shape)
ax.scatter(best_idx[1], best_idx[0], marker="*", s=300, color="gold",
           zorder=5, label=f"Best correlation: Δlon={shift_range[best_idx[1]]:+d}°, Δlat={shift_range[best_idx[0]]:+d}°"
                           f"\nR={r_spatial[best_idx]:.3f}\nR²={r_spatial[best_idx]**2:.3f}" )
ax.legend(fontsize=10)
plt.tight_layout()
plt.savefig(f'/home/oscholz/plots/{folder}/spatial_crosscorr_DoF_{data}.png',
            dpi=150, bbox_inches="tight")
plt.show()

print(f"Bester Versatz: Δlon={shift_range[best_idx[1]]:+d}°, Δlat={shift_range[best_idx[0]]:+d}°")
print(f"r = {r_spatial[best_idx]:.3f}")
```

```python
# ── Besten Versatz aus räumlicher Kreuzkorrelation verwenden ──────────────────
best_d_lat = shift_range[best_idx[0]]
best_d_lon = shift_range[best_idx[1]]

# Dof-Feld mit bestem Versatz verschieben
dof_shifted_best = np.roll(np.roll(dof_vals_amz, best_d_lat, axis=0), best_d_lon, axis=1)
area_shifted_best = np.roll(np.roll(dof_km2_vals_amz, best_d_lat, axis=0), best_d_lon, axis=1)

# Randmaske
mask_best = np.ones_like(dof_vals_amz, dtype=bool)
if best_d_lat > 0:
    mask_best[:best_d_lat, :] = False
elif best_d_lat < 0:
    mask_best[best_d_lat:, :] = False
if best_d_lon > 0:
    mask_best[:, :best_d_lon] = False
elif best_d_lon < 0:
    mask_best[:, best_d_lon:] = False

valid_best = mask_best & ~np.isnan(dof_shifted_best) & ~np.isnan(trend_vals_amz)

dof_flat_best   = dof_shifted_best[valid_best].flatten()
area_flat_best  = area_shifted_best[valid_best].flatten()
trend_flat_best = trend_vals_amz[valid_best].flatten()

# ── Lineare Regression ────────────────────────────────────────────────────────
slope_b, intercept_b, r_value_b, p_value_b, std_err_b = stats.linregress(dof_flat_best, trend_flat_best)
slope_c, intercept_c, r_value_c, p_value_c, std_err_c = stats.linregress(area_flat_best, trend_flat_best)

# ── Plot ──────────────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(dof_flat_best, trend_flat_best, alpha=0.6, edgecolors="k",
           linewidths=0.3, s=40, color="steelblue")
x_line_b = np.linspace(dof_flat_best.min(), dof_flat_best.max(), 100)
y_line_b = slope_b * x_line_b + intercept_b
ax.plot(x_line_b, y_line_b, color="red", linewidth=2,
        label=f"Slope: {slope_b:.4f}\n$R^2$ = {r_value_b**2:.3f}\np = {p_value_b:.4f}")
ax.axhline(0, color="gray", linewidth=0.8, linestyle="--")
ax.set_xlabel("Difference of Forrest (Pixel Count)", fontsize=12)
ax.set_ylabel(f"{header} trend (per decade)", fontsize=12)
ax.set_title(f"DoF & {header} trend (low clouds) – best spatial lag\n"
             f"Δlat={best_d_lat:+d}°, Δlon={best_d_lon:+d}° | "
             f"Amazon (14°S–7°N, 80°W–40°W), only significant trends (p < 0.05)", fontsize=12)
ax.legend(fontsize=11)
plt.tight_layout()
plt.savefig(f'/home/oscholz/plots/{folder}/scatter_DoF_{data}_bestlag.png',
            dpi=150, bbox_inches="tight")
plt.show()

# ── Plot für Area ──────────────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(area_flat_best, trend_flat_best, alpha=0.6, edgecolors="k",
           linewidths=0.3, s=40, color="steelblue")
x_line_c = np.linspace(area_flat_best.min(), area_flat_best.max(), 100)
y_line_c = slope_c * x_line_c + intercept_c
ax.plot(x_line_c, y_line_c, color="red", linewidth=2,
        label=f"Slope: {slope_b:.4f}\n$R^2$ = {r_value_b**2:.3f}\np = {p_value_b:.4f}")
ax.axhline(0, color="gray", linewidth=0.8, linestyle="--")
ax.set_xlabel("Lost forrest Area", fontsize=12)
ax.set_ylabel(f"{header} trend (per decade)", fontsize=12)
ax.set_title(f" Lost forest area& {header} trend (low clouds) – best spatial lag\n"
             f"Δlat={best_d_lat:+d}°, Δlon={best_d_lon:+d}° | "
             f"Amazon (14°S–7°N, 80°W–40°W), only significant trends (p < 0.05)", fontsize=12)
ax.legend(fontsize=11)
plt.tight_layout()
plt.savefig(f'/home/oscholz/plots/{folder}/scatter_area_{data}_bestlag.png',
            dpi=150, bbox_inches="tight")
plt.show()
```
